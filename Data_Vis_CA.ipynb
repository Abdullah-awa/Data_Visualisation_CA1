```python

df = pd.read_excel("all_data_M_2023.xlsx")

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Rows:")
display(df.head())
```

```python
print("Dataset Information:")
print("--------------------")

print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
# Create a copy of the original dataset
df_clean = df.copy()

# Convert numerical columns into numeric format
numeric_columns = [
    'TOT_EMP',
    'EMP_PRSE',
    'JOBS_1000',
    'LOC_QUOTIENT',
    'PCT_TOTAL',
    'PCT_RPT',
    'H_MEAN',
    'A_MEAN',
    'MEAN_PRSE',
    'H_PCT10',
    'H_PCT25',
    'H_MEDIAN',
    'H_PCT75',
    'H_PCT90',
    'A_PCT10',
    'A_PCT25',
    'A_MEDIAN',
    'A_PCT75',
    'A_PCT90',
    'ANNUAL',
    'HOURLY'
]

for col in numeric_columns:
    df_clean[col] = pd.to_numeric(
        df_clean[col], errors='coerce'
    )

print("Data cleaning completed!")

print("\nUpdated Data Types:")
print(df_clean[numeric_columns].dtypes)

print("\nMissing Values After Cleaning:")
print(df_clean[numeric_columns].isnull().sum())
```

```python
# Find relevant occupations

keywords = (
    'Data Scientist|Artificial Intelligence|'
    'Software Developer|Software Engineer|'
    'Machine Learning|Data Engineer'
)

relevant_occupations = df_clean[
    df_clean['OCC_TITLE'].str.contains(
        keywords,
        case=False,
        na=False
    )
]

print("Relevant Occupations Found:")
print(
    relevant_occupations['OCC_TITLE']
    .value_counts()
)

print("\nTotal Records:")
print(len(relevant_occupations))
```

```python
# Analyse relevant occupations

occupation_summary = relevant_occupations.groupby('OCC_TITLE').agg(
    Total_Employment=('TOT_EMP', 'sum'),
    Average_Annual_Salary=('A_MEAN', 'mean')
).reset_index()

print("Occupation Analysis:")
display(occupation_summary)
```

```python
# Compare Employment and Salary

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.barplot(
    data=occupation_summary,
    x='OCC_TITLE',
    y='Total_Employment',
    ax=axes[0]
)

axes[0].set_title('Total Employment by Occupation')
axes[0].set_xlabel('Occupation')
axes[0].set_ylabel('Total Employment')
axes[0].tick_params(axis='x', rotation=15)

sns.barplot(
    data=occupation_summary,
    x='OCC_TITLE',
    y='Average_Annual_Salary',
    ax=axes[1]
)

axes[1].set_title('Average Annual Salary by Occupation')
axes[1].set_xlabel('Occupation')
axes[1].set_ylabel('Average Salary ($)')
axes[1].tick_params(axis='x', rotation=15)

plt.tight_layout()
plt.show()
```

The first bar graph contrasts the overall employment of software developers with data scientists. According to the findings, there are roughly 11.67 million software developers employed overall, compared to 2.42 million data scientists. This suggests that there are more jobs in software development in the dataset.

```python
# Salary Distribution Analysis

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=relevant_occupations,
    x='OCC_TITLE',
    y='A_MEAN'
)

plt.title('Salary Distribution by Occupation')
plt.xlabel('Occupation')
plt.ylabel('Annual Mean Salary ($)')
plt.xticks(rotation=15)

plt.tight_layout()
plt.show()
```

### Distribution of Salaries by Occupation

The annual mean wage distribution of software developers and data scientists is contrasted in the boxplot.

According to the graph, the median compensation for software developers is higher than that of data scientists. Additionally, the pay distribution of software developers is broader, suggesting that incomes vary more. There are some outliers in both professions, which indicate abnormally high or low wage values.

Although income levels differ between the two professions, the boxplot shows that software developers often earn greater salaries in this dataset.
