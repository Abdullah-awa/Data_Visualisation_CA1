```python
# Create a copy of the original dataset
df_clean = df.copy()

# Convert numerical columns into numeric format
numeric_columns = [
    'TOT_EMP',
    'EMP_PRSE',
    'JOBS_1000',
    'LOC_QUOTIENT',
    'PCT_TOTAL',
    'PCT_RPT',
    'H_MEAN',
    'A_MEAN',
    'MEAN_PRSE',
    'H_PCT10',
    'H_PCT25',
    'H_MEDIAN',
    'H_PCT75',
    'H_PCT90',
    'A_PCT10',
    'A_PCT25',
    'A_MEDIAN',
    'A_PCT75',
    'A_PCT90',
    'ANNUAL',
    'HOURLY'
]

for col in numeric_columns:
    df_clean[col] = pd.to_numeric(
        df_clean[col], errors='coerce'
    )

print("Data cleaning completed!")

print("\nUpdated Data Types:")
print(df_clean[numeric_columns].dtypes)

print("\nMissing Values After Cleaning:")
print(df_clean[numeric_columns].isnull().sum())
```

```python
# Find relevant occupations

keywords = (
    'Data Scientist|Artificial Intelligence|'
    'Software Developer|Software Engineer|'
    'Machine Learning|Data Engineer'
)

relevant_occupations = df_clean[
    df_clean['OCC_TITLE'].str.contains(
        keywords,
        case=False,
        na=False
    )
]

print("Relevant Occupations Found:")
print(
    relevant_occupations['OCC_TITLE']
    .value_counts()
)

print("\nTotal Records:")
print(len(relevant_occupations))
```

```python
# Analyse relevant occupations

occupation_summary = relevant_occupations.groupby('OCC_TITLE').agg(
    Total_Employment=('TOT_EMP', 'sum'),
    Average_Annual_Salary=('A_MEAN', 'mean')
).reset_index()

print("Occupation Analysis:")
display(occupation_summary)
```
