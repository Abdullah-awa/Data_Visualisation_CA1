```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python
df = pd.read_excel("all_data_M_2023.xlsx")
```

```python

df = pd.read_excel("all_data_M_2023.xlsx")

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Rows:")
display(df.head())
```

```python
print("Dataset Information:")
print("--------------------")

print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
# Create a copy of the original dataset
df_clean = df.copy()

# Convert numerical columns into numeric format
numeric_columns = [
    'TOT_EMP',
    'EMP_PRSE',
    'JOBS_1000',
    'LOC_QUOTIENT',
    'PCT_TOTAL',
    'PCT_RPT',
    'H_MEAN',
    'A_MEAN',
    'MEAN_PRSE',
    'H_PCT10',
    'H_PCT25',
    'H_MEDIAN',
    'H_PCT75',
    'H_PCT90',
    'A_PCT10',
    'A_PCT25',
    'A_MEDIAN',
    'A_PCT75',
    'A_PCT90',
    'ANNUAL',
    'HOURLY'
]

for col in numeric_columns:
    df_clean[col] = pd.to_numeric(
        df_clean[col], errors='coerce'
    )

print("Data cleaning completed!")

print("\nUpdated Data Types:")
print(df_clean[numeric_columns].dtypes)

print("\nMissing Values After Cleaning:")
print(df_clean[numeric_columns].isnull().sum())
```

```python
# Find relevant occupations

keywords = (
    'Data Scientist|Artificial Intelligence|'
    'Software Developer|Software Engineer|'
    'Machine Learning|Data Engineer'
)

relevant_occupations = df_clean[
    df_clean['OCC_TITLE'].str.contains(
        keywords,
        case=False,
        na=False
    )
]

print("Relevant Occupations Found:")
print(
    relevant_occupations['OCC_TITLE']
    .value_counts()
)

print("\nTotal Records:")
print(len(relevant_occupations))
```

```python
# Analyse relevant occupations

occupation_summary = relevant_occupations.groupby('OCC_TITLE').agg(
    Total_Employment=('TOT_EMP', 'sum'),
    Average_Annual_Salary=('A_MEAN', 'mean')
).reset_index()

print("Occupation Analysis:")
display(occupation_summary)
```

```python
# Compare Employment and Salary

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.barplot(
    data=occupation_summary,
    x='OCC_TITLE',
    y='Total_Employment',
    ax=axes[0]
)

axes[0].set_title('Total Employment by Occupation')
axes[0].set_xlabel('Occupation')
axes[0].set_ylabel('Total Employment')
axes[0].tick_params(axis='x', rotation=15)

sns.barplot(
    data=occupation_summary,
    x='OCC_TITLE',
    y='Average_Annual_Salary',
    ax=axes[1]
)

axes[1].set_title('Average Annual Salary by Occupation')
axes[1].set_xlabel('Occupation')
axes[1].set_ylabel('Average Salary ($)')
axes[1].tick_params(axis='x', rotation=15)

plt.tight_layout()
plt.show()
```

The first bar graph contrasts the overall employment of software developers with data scientists. According to the findings, there are roughly 11.67 million software developers employed overall, compared to 2.42 million data scientists. This suggests that there are more jobs in software development in the dataset.

### Objective
The objective of these bar charts is to compare total employment and average annual salaries between Data Scientists and Software Developers.

### Why This Visualisation Was Chosen
Bar charts were selected because they make it easy to compare numerical values across different occupation categories. They provide a clear visual comparison of employment levels and average salaries.

### Findings
The analysis shows that Software Developers have higher total employment than Data Scientists. Software Developers also have a higher average annual salary of approximately $117,651, compared  with  approximately $103,932 for Data Scientists.

```python
# Salary Distribution Analysis

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=relevant_occupations,
    x='OCC_TITLE',
    y='A_MEAN'
)

plt.title('Salary Distribution by Occupation')
plt.xlabel('Occupation')
plt.ylabel('Annual Mean Salary ($)')
plt.xticks(rotation=15)

plt.tight_layout()
plt.show()
```

### Distribution of Salaries by Occupation

The annual mean wage distribution of software developers and data scientists is contrasted in the boxplot.

According to the graph, the median compensation for software developers is higher than that of data scientists. Additionally, the pay distribution of software developers is broader, suggesting that incomes vary more. There are some outliers in both professions, which indicate abnormally high or low wage values.

Although income levels differ between the two professions, the boxplot shows that software developers often earn greater salaries in this dataset.

### Objective
The objective of this visualisation is to compare the salary distributions of Data Scientists and Software Developers.

### Why This Visualisation Was Chosen
A boxplot was selected because it displays the median, quartiles, overall spread and potential outliers in salary data. This makes it useful for comparing salary distributions between occupations.

### Findings
The boxplot indicates that Software Developers have a higher median salary than Data Scientists. It also shows differences in salary variation and the presence of potential outliers in both occupations.

```python
# Calculate Salary Statistics

salary_statistics = relevant_occupations.groupby('OCC_TITLE')['A_MEAN'].agg(
    ['count', 'mean', 'median', 'min', 'max', 'std']
).round(2)

print("Salary Statistics by Occupation:")
display(salary_statistics)
```

```python
 #Filter Data Scientists employment by US industries

industry_data = relevant_occupations[
    (relevant_occupations['OCC_TITLE'] == 'Data Scientists') &
    (relevant_occupations['AREA_TITLE'] == 'U.S.')
].copy()

industry_data = industry_data[
    industry_data['NAICS_TITLE'].notna()
]

industry_data['TOT_EMP'] = pd.to_numeric(
    industry_data['TOT_EMP'], errors='coerce'
)

industry_data = industry_data.dropna(subset=['TOT_EMP'])

# Remove overall cross-industry categories
industry_data = industry_data[
    ~industry_data['NAICS_TITLE'].str.contains(
        'Cross-industry', case=False, na=False
    )
]

industry_data = industry_data[
    ['NAICS_TITLE', 'TOT_EMP']
].drop_duplicates()

industry_data = industry_data.sort_values(
    by='TOT_EMP',
    ascending=False
)

display(industry_data.head(10))
```

```python
# Visualise Top 10 Industries Employing Data Scientists

top_10_industries = industry_data.head(10)

plt.figure(figsize=(12, 7))

plt.barh(
    top_10_industries['NAICS_TITLE'],
    top_10_industries['TOT_EMP']
)

plt.title('Top 10 Industries Employing Data Scientists')
plt.xlabel('Total Employment')
plt.ylabel('Industry')

plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```

### Examining the Top 10 Sectors

The top ten industries in the US that employ data scientists are displayed in the bar chart.

With 60,800 Data Scientists, Professional, Scientific, and Technical Services has the largest workforce. With 33,620 workers, the second-largest industry is finance and insurance.

With 23,150 workers, Computer Systems Design and Related Services offers substantial job prospects.

Overall, the graph shows that professional services, finance, and technology-related industries employ the majority of data scientists.

### Objective
The objective of this visualisation is to identify the top 10 industries employing Data Scientists.

### Why This Visualisation Was Chosen
A horizontal bar chart was selected because it allows easy comparison between industries. It also makes long industry names easier to read.

### Findings
The results show that Professional, Scientific, and Technical Services employ the highest number of Data Scientists. Finance and Insurance ranks second, followed by Computer Systems Design and Related Services.

```python
# Pie Chart: Employment Share by Occupation

plt.figure(figsize=(8, 8))

plt.pie(
    occupation_summary['Total_Employment'],
    labels=occupation_summary['OCC_TITLE'],
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Employment Share by Occupation')
plt.axis('equal')
plt.show()
```

### Pie Chart: Employment Share by Occupation

**Objective:**  
The objective of this visualisation is to compare the employment share of Data Scientists and Software Developers in the selected dataset.

**Why this visualisation was chosen:**  
A pie chart is suitable for showing proportions and percentages of a whole. It makes it easy to understand how total employment is divided between the two occupations.

**What was learned:**  
The chart shows that Software Developers account for a much larger share of employment than Data Scientists in the selected records. This indicates that Software Development has a higher employment level within these selected occupations.

```python
# Histogram: Annual Salary Distribution by Occupation

plt.figure(figsize=(10, 6))

for occupation in relevant_occupations['OCC_TITLE'].unique():
    salary_data = relevant_occupations[
        relevant_occupations['OCC_TITLE'] == occupation
    ]['A_MEAN'].dropna()

    plt.hist(
        salary_data,
        bins=20,
        alpha=0.5,
        label=occupation
    )

plt.title('Annual Salary Distribution by Occupation')
plt.xlabel('Annual Salary ($)')
plt.ylabel('Frequency')
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.show()
```

### Histogram: Annual Salary Distribution by Occupation

**Objective:**  
The objective of this visualisation is to examine and compare the distribution of annual salaries for Data Scientists and Software Developers.

**Why this visualisation was chosen:**  
A histogram is suitable for displaying the frequency distribution of numerical data. It helps identify salary ranges, patterns and variations between the two occupations.

**What was learned:**  
The histogram shows how annual salaries are distributed across both occupations. It helps identify common salary ranges and shows the variation in earnings between Data Scientists and Software Developers.
